```python
import gc

class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"object {self.name} created")

    def __del__(self):
        print(f"object {self.name}  deleted")


obj1=MyObject("obj1")
obj2=MyObject("obj2")
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2

## manually trigger the garbage collection
gc.collect()



```

```python
def  generater_numbers(n):
    for i in range(n):
        yield i

for num in generater_numbers(100):
    print(num)
    if num>=10:
        break
```
