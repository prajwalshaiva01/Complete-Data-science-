```python
import numpy as np
import pandas as pd

np.random.seed(123)

n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio)
n_class_1=n_samples-n_class_0
```

```python
n_class_0,n_class_1
```

```python
## crete my dataframe
## Upsampled
class_0=pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1=pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
df_minoruty=df[df['target']==1]
df_minoruty=df[df['target']==0]
```

```python
from sklearn.utils import resample
```

```python
df_minoruty_upsampled=resample(df_minoruty,replace=True,
         n_samples=len(df_minoruty),
         random_state=42
         )
```

```python
df_minoruty_upsampled.shape
```

```python
df_minoruty_upsampled.head()
```

```python
df_upsampled=pd.concat([df_minoruty,df_minoruty_upsampled])
```

```python
df_minoruty_upsampled['target'].value_counts()
```

```python
## down sampled

## crete my dataframe

class_0=pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1=pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
df=pd.concat([class_0,class_1]).reset_index(drop=True)
print(df['target'].value_counts())
```

```python
df_minoruty=df[df['target']==1]
df_minoruty=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majoruty_downsampled=resample(df_majoruty,replace=True,
         n_samples=len(df_majoruty),
         random_state=42
         )
```

```python
df_majoruty_downsampled.head()
```

```python
df_downampled=pd.concat([df_minoruty,df_minoruty_upsampled])
```
