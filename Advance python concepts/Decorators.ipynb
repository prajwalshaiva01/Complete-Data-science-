```python
## Decorator
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advance python cource")
        func()
        print("Please  learn these concepts properly")
    return sub_welcome_method()
```

```python
def cource_introduction():
    print("THis is an advanaced python course")
cource_introduction()
```

```python
main_welcome(cource_introduction)
```

```python
@main_welcome
def cource_introduction():
    print("THis is an advanaced python course")
```

```python
## Decorator
def my_decorator(func):
    def wrapper():
        print("something is happening before the funcation is called.")
        func()
        print("something is happening ofter  the funcation is called. ")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("hello")
```

```python
say_hello()
```

```python
## Decorators with argument
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
