```python
import sqlite3
```

```python
## connect to an sqlite data base
coenction=sqlite3.connect('example.db')
coenction
```

```python
cursor=coenction.cursor()
```

```python
## create a table
cursor.execute('''
Create Table If Not Exists Employees(
    id Integer Primary key,
    name Text Not Null,
    age Integer,
    department Text
    )
''')

## commit the changes
coenction.commit()
```

```python
cursor.execute('''
select * from employees
''')
```

```python
## insert the data in sqlite 

cursor.execute('''
Insert into employees(name,age,department)
             values('Prajwal',23,'MCA')
''')

cursor.execute('''
Insert into employees(name,age,department)
             values('Darshan',23,'BCA')
''')

cursor.execute('''
Insert into employees(name,age,department)
             values('Kiran',23,'CA')
''')

cursor.execute('''
Insert into employees(name,age,department)
             values('Chandu',24,'MA')
''')

cursor.execute('''
Insert into employees(name,age,department)
             values('Raju',22,'MCA')
''')

coenction.commit()
```

```python
## query  the data from the table
cursor.execute('select * from employees')
rows=cursor.fetchall()

## print the qurey
for row in rows:
    print(row)
```

```python
## update the data 
cursor.execute('''
UPDATE employees
Set age=24
where name="Prajwal"
''')
coenction.commit()
```

```python
## query  the data from the table
cursor.execute('select * from employees')
rows=cursor.fetchall()

## print the qurey
for row in rows:
    print(row)
```

```python
### working with selases data 

coenction=sqlite3.connect('sales_data.db')
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INTEGER PRIMARY KEY,
    date TEXT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
    )
''')

## insert data correctly with commas
sales_data = [
    ('2023-01-01', 'product1', 100, 'north'),
    ('2023-01-02', 'product2', 200, 'south'),
    ('2023-01-03', 'product3', 150, 'east'),
    ('2023-01-04', 'product4', 250, 'west')
]
cursor.executemany('''
Insert into sales(date, product, sales, region)
values(?,?,?,?)
''', sales_data)

coenction.commit()
```

```python
cursor.execute('SELECT * FROM sales')
rows=cursor.fetchall()

for row in rows:
    print(row)
```
