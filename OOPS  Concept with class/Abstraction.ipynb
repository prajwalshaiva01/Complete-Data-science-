```python
from abc import ABC,abstractmethod

## abstract class
class Vehicle(ABC):
    def drive(self):
        print("the vehical is used for driving")

    @abstractmethod
    def start_engine(self):
        pass 

class Car(Vehicle):
    def start_engine(self):
        print("car engin started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car=Car()
operate_vehicle(car)

```
