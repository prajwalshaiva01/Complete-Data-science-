```python
class Person:
    def __init__(self,name,age,gender):
        self.__name=name # privite var
        self.__age=age  ## privite var
        self.gender=gender

def get_name(person):
    return person.__name

person=Person("prajwal",23,"male")
get_name(person)
```

```python
class Person:
    def __init__(self,name,age,gender):
        self._name=name # protected var
        self._age=age  ## protected var
        self.gender=gender

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)
employee=Employee("prajwal",23,"male")
print(employee._name)
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self._age = age

    # setter method for name
    def set_name(self, name):
        self.__name = name

    # getter method for name
    def get_name(self):
        return self.__name

    # setter method for age
    def set_age(self, age):
        if age > 0:
            self._age = age
        else:
            print("age cannot be -ve")

    # getter method for age
    def get_age(self):
        return self._age

# Object creation and testing
person = Person("prajwal", 23)
print(person.get_name())
print(person.get_age())

person.set_age(24)
print(person.get_age())

person.set_age(-3)
```
