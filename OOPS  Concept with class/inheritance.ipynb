```python
## inheritance (single inheritance)
class car:
    def __init__(self,windows,doors,engintype):
        self.windows=windows 
        self.doors=doors
        self.engintype=engintype
    def drive(self):
        print(f"the person will drive the {self.engintype} car")
```

```python
car1=car(4,5,"petrol")
print(car1.drive())
```

```python
car1.drive()
```

```python
class Tesla(car):
    def __init__(self, windows, doors, engintype,selfdriving):
        super().__init__(windows,doors,engintype)
        self.is_selfdriving=selfdriving
    def selfdriving(self):
        print(f"Tesla supoorts self driving :{self.is_selfdriving}")
```

```python
tesla1=Tesla(4,5,"electric",True)
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
 ## multiple inheritance
## base class 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print(f"sub class implement this mothod")
## base class 2

class pet:
    def __init__(self,owner):
        self.owner=owner

## derive class
class Dog(Animal,pet):
    def __init__(self, name,owner):
        Animal.__init__(self,name)
        pet.__init__(self,owner)

    def speak(self):
        return f"{self.name} say woof"

    
##  create an object
dog=Dog("buddy","prajwal")
print(dog.speak())
print(f"owne:{dog.owner}")
```
