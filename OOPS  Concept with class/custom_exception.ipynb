```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year=int(input("enter the dob"))
age=2024-year
try:

    if age<=30 and  age>=20:
        print("the age is valid")
    else:
        raise dobException
except dobException:
    print("sorry, your age should be greter then 20 or less then 30")
    
```
